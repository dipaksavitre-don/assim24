```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load datasets
classification_df = pd.read_csv("Session24_Classification_Dataset.csv")
regression_df = pd.read_csv("Session24_Regression_Dataset.csv")

# Display datasets
print("Classification Dataset:")
display(classification_df.head())

print("\nRegression Dataset:")
display(regression_df.head())

# Check missing values
print("\nClassification Missing Values:")
print(classification_df.isnull().sum())

print("\nRegression Missing Values:")
print(regression_df.isnull().sum())

# Classification X and y
X_class = classification_df.drop("Risk", axis=1)
y_class = classification_df["Risk"]

# Regression X and y
X_reg = regression_df.drop("Price", axis=1)
y_reg = regression_df["Price"]

# Train-test split
X_class_train, X_class_test, y_class_train, y_class_test = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42, stratify=y_class
)

X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# Scaling
class_scaler = StandardScaler()
X_class_train_scaled = class_scaler.fit_transform(X_class_train)
X_class_test_scaled = class_scaler.transform(X_class_test)

reg_scaler = StandardScaler()
X_reg_train_scaled = reg_scaler.fit_transform(X_reg_train)
X_reg_test_scaled = reg_scaler.transform(X_reg_test)

print("\nClassification X:", X_class.shape)
print("Classification y:", y_class.shape)

print("Regression X:", X_reg.shape)
print("Regression y:", y_reg.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

# Create models
classification_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": SVC(),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB()
}

classification_results = []

for name, model in classification_models.items():

    # Train
    model.fit(X_class_train_scaled, y_class_train)

    # Prediction
    y_pred = model.predict(X_class_test_scaled)

    # Metrics
    accuracy = accuracy_score(y_class_test, y_pred)
    precision = precision_score(y_class_test, y_pred)
    recall = recall_score(y_class_test, y_pred)
    f1 = f1_score(y_class_test, y_pred)

    classification_results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("\nAccuracy:")
    print(accuracy)

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_class_test, y_pred))

    print("\nClassification Report:")
    print(classification_report(y_class_test, y_pred))

# Comparison table
classification_comparison = pd.DataFrame(
    classification_results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

print("\nClassification Model Comparison:")
display(classification_comparison)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score

# Create regression models
regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "SVR": SVR(),
    "KNN Regressor": KNeighborsRegressor(n_neighbors=5)
}

regression_results = []

for name, model in regression_models.items():

    # Train
    model.fit(X_reg_train_scaled, y_reg_train)

    # Prediction
    y_pred = model.predict(X_reg_test_scaled)

    # R2 Score
    r2 = r2_score(y_reg_test, y_pred)

    regression_results.append([
        name,
        r2
    ])

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("R² Score:", r2)

# Comparison table
regression_comparison = pd.DataFrame(
    regression_results,
    columns=["Model", "R2 Score"]
)

print("\nRegression Model Comparison:")
display(regression_comparison)

# Best regression model
best_regression = regression_comparison.loc[
    regression_comparison["R2 Score"].idxmax()
]

print("\nBest Regression Model:")
print(best_regression)
```

```python
import joblib

# -----------------------------
# Find best classification model
# -----------------------------

best_classification_name = classification_comparison.loc[
    classification_comparison["Accuracy"].idxmax(),
    "Model"
]

best_classification_model = classification_models[
    best_classification_name
]

print("Best Classification Model:")
print(best_classification_name)

# Save classification model
joblib.dump(
    best_classification_model,
    "best_classification_model.pkl"
)

# Save classification scaler
joblib.dump(
    class_scaler,
    "classification_scaler.pkl"
)

# -----------------------------
# Find best regression model
# -----------------------------

best_regression_name = regression_comparison.loc[
    regression_comparison["R2 Score"].idxmax(),
    "Model"
]

best_regression_model = regression_models[
    best_regression_name
]

print("\nBest Regression Model:")
print(best_regression_name)

# Save regression model
joblib.dump(
    best_regression_model,
    "best_regression_model.pkl"
)

# Save regression scaler
joblib.dump(
    reg_scaler,
    "regression_scaler.pkl"
)

print("\nAll models and scalers saved successfully!")
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

st.title("Session 24 - Multi Model Prediction App")

problem = st.selectbox(
    "Choose Problem Type",
    ["Classification", "Regression"]
)

if problem == "Classification":

    model = joblib.load("best_classification_model.pkl")
    scaler = joblib.load("classification_scaler.pkl")

    st.subheader("Classification Input")

    age = st.number_input("Age", 18, 100, 30)
    bmi = st.number_input("BMI", 10.0, 50.0, 25.0)
    bp = st.number_input("Blood Pressure", 80, 200, 120)
    glucose = st.number_input("Glucose", 50, 300, 100)
    cholesterol = st.number_input("Cholesterol", 100, 400, 180)
    exercise = st.number_input("Exercise Hours", 0.0, 20.0, 5.0)

    if st.button("Predict"):

        input_data = pd.DataFrame([[
            age,
            bmi,
            bp,
            glucose,
            cholesterol,
            exercise
        ]], columns=[
            "Age",
            "BMI",
            "BloodPressure",
            "Glucose",
            "Cholesterol",
            "ExerciseHours"
        ])

        input_scaled = scaler.transform(input_data)

        prediction = model.predict(input_scaled)

        if prediction[0] == 1:
            st.error("Prediction: High Risk")
        else:
            st.success("Prediction: Low Risk")


else:

    model = joblib.load("best_regression_model.pkl")
    scaler = joblib.load("regression_scaler.pkl")

    st.subheader("Regression Input")

    area = st.number_input("Area (Sq Ft)", 300, 10000, 1500)
    bedrooms = st.number_input("Bedrooms", 1, 10, 3)
    bathrooms = st.number_input("Bathrooms", 1, 10, 2)
    house_age = st.number_input("House Age", 0, 100, 10)
    distance = st.number_input("Distance From City", 0.0, 100.0, 10.0)
    parking = st.number_input("Parking Spaces", 0, 10, 1)

    if st.button("Predict Price"):

        input_data = pd.DataFrame([[
            area,
            bedrooms,
            bathrooms,
            house_age,
            distance,
            parking
        ]], columns=[
            "AreaSqFt",
            "Bedrooms",
            "Bathrooms",
            "HouseAge",
            "DistanceFromCity",
            "ParkingSpaces"
        ])

        input_scaled = scaler.transform(input_data)

        prediction = model.predict(input_scaled)

        st.success(
            f"Predicted House Price: ₹{prediction[0]:,.2f}"
        )
```
